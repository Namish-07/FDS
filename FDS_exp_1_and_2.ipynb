```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("titanic.csv")

print("First 5 Records")
print(df.head())

print("\nDataset Shape")
print(df.shape)

print("\nColumn Names")
print(df.columns)

print("\nData Types")
print(df.dtypes)

print("\nMissing Values")
print(df.isnull().sum())

print("\nStatistical Summary")
print(df.describe())

# ==========================================
# Exploratory Data Analysis (EDA)
# ==========================================

# Survival Count
sns.countplot(x='Survived', data=df)
plt.title("Survival Count")
plt.show()

# Gender Count
sns.countplot(x='Sex', data=df)
plt.title("Gender Distribution")
plt.show()

# Passenger Class
sns.countplot(x='Pclass', data=df)
plt.title("Passenger Class")
plt.show()

# Age Distribution
plt.figure(figsize=(6,4))
sns.histplot(df['Age'], bins=20, kde=True)
plt.title("Age Distribution")
plt.show()

# Fare Distribution
plt.figure(figsize=(6,4))
sns.histplot(df['Fare'], bins=20, kde=True)
plt.title("Fare Distribution")
plt.show()

# Correlation Heatmap
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap='coolwarm')

plt.title("Correlation Matrix")
plt.show()

# ==========================================
# (b) Identify Attributes
# ==========================================

print("\nNumeric Attributes")
numeric = df.select_dtypes(include=['int64','float64']).columns
print(numeric)

print("\nNominal Attributes")
nominal = ['Sex','Names']

print(nominal)

print("\nBinary Attributes")
binary = ['Survived']

print(binary)

# ==========================================
# (c) Dissimilarity for Nominal Attributes
# Example:
# Sex and Embarked
# Simple Matching Dissimilarity
# ==========================================

sex = df['Sex']
name = df['Name']

count = 0
total = 0

for s,e in zip(sex,name):

    if pd.isna(s) or pd.isna(e):
        continue

    total += 1

    if s != e:
        count += 1

dissimilarity = count/total

print("\nNominal Attribute Dissimilarity")
print("Sex vs Name =", dissimilarity)

# ==========================================
# (d) Numeric Attribute Dissimilarity
# Example:
# Age and Fare
# Euclidean Distance
# ==========================================

# Remove Missing Values

temp = df[['Age','Fare']].dropna()

age = temp.iloc[0]['Age']
fare = temp.iloc[0]['Fare']

age2 = temp.iloc[1]['Age']
fare2 = temp.iloc[1]['Fare']

distance = np.sqrt((age-age2)**2 + (fare-fare2)**2)

print("\nEuclidean Distance")
print(distance)

# ==========================================
# Nominal Attribute Similarity
# Simple Matching Similarity
# ==========================================

sex = df['Sex']
name = df['Name']

match = 0
total = 0

for s, n in zip(sex, name):

    if pd.isna(s) or pd.isna(n):
        continue

    total += 1

    if s == n:
        match += 1

similarity = match / total

print("\nNominal Attribute Similarity")
print("Sex vs Name Similarity =", similarity)

euclidean_similarity = 1 / (1 + distance)

print("\nEuclidean Similarity")
print(euclidean_similarity)

# ==========================================
# Manhattan Distance
# ==========================================

manhattan = abs(age-age2)+abs(fare-fare2)

print("\nManhattan Distance")
print(manhattan)

manhattan_similarity = 1 / (1 + manhattan)

print("\nManhattan Similarity")
print(manhattan_similarity)

# ==========================================
# Minkowski Distance (p=3)
# ==========================================

minkowski=((abs(age-age2)**3)+(abs(fare-fare2)**3))**(1/3)

print("\nMinkowski Distance")
print(minkowski)

minkowski_similarity = 1 / (1 + minkowski)

print("\nMinkowski Similarity")
print(minkowski_similarity)



```
